```python
import pyspark
from pyspark.sql import SparkSession
from pyspark.sql.types import *
```

```python
spark = SparkSession.builder.getOrCreate()
```

```python
df = spark.read.csv('videos-stats.csv')
```

```python
df.take(8)
```

```python
df.printSchema()
```

```python
df = spark.read.option('inferSchema', 'true').csv('videos-stats.csv')
```

```python
df.write.mode('overwrite').option('header', 'true').parquet('videos-parquet')
```

```python
df.show()
```

```python
df = spark.read.parquet('videos-parquet')
```

```python
df.write.option('header', 'true').saveAsTable('tb_videos')
```

```python
spark.catalog.listTables()
```

```python
tab_df = spark.sql('select * from tb_videos')

tab_df.show()
```
